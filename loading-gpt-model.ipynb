Importing the required libraries, transformers for the model and the tokenizer and tensorflow as the 
backend framework

```python
# importing the necessary libraries
import tensorflow as tf
from transformers import TFGPT2LMHeadModel, GPT2Tokenizer
```

```python
# checking for gpu availabilty
print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))
```

#### Loading pretrained model and tokenizer
- **Tokenizer:** Converts text into numbers the model understands.
- **Model:** The actual neural network (GPT-2) trained on massive amounts of data.

```python
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = TFGPT2LMHeadModel.from_pretrained('gpt2')
```

##### defining a generatiom function
**the function takes a text prompt, encodes it, feed it to the model and decodes the result back into text**

```python
def generate_text(prompt, max_length=100):
    #encoding the prompt
    input_ids = tokenizer.encode(prompt, return_tensors='tf')
    #generating text
    output = model.generate(input_ids, max_length=max_length, num_return_sequences=1, 
    no_repeat_ngram_size=2, do_sample=True, top_k=50)
    #decoding the generated text
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    
    return generated_text
```

##### testing model to generate text

```python
#first prompt
prompt1 = "I love reading novels because they make me" 
generated_text = generate_text(prompt1)
print(generated_text)
```

```python
#Second prompt
prompt2 = "Reading books and listening to music is" 
generated_text = generate_text(prompt2)
print(generated_text)
```
