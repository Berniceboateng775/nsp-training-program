Create a histogram of a dataset of your choosing using matplotlib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import cufflinks as cf
from IPython.display import display, HTML

cf.set_config_file(sharing='public', theme='ggplot', offline=True)
cf.getThemes()
```

```python
cleaned_laptop = pd.read_csv(r"C:\Users\BerniceBoateng\Desktop\machine-learning-zoomcamp\cleaned_laptop_data.csv")
cleaned_laptop.head()
```

```python
#histogram of the price column
cleaned_laptop['price'].plot(kind='hist', bins=20, figsize=(7, 4), title='Histogram of Laptop Prices', color='blue')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
#histogram of the inches column
cleaned_laptop['inches'].plot(kind='hist', bins=20, figsize=(7, 4), title='Histogram of Laptop Inches', color='red')
plt.xlabel('Inches')
plt.ylabel('Frequency')
plt.show()
```

**Question 2**                                                                           
Perform basic data visualization using matplotlib and seaborn on your selected dataset. Create and Interpret a corelation matrix

```python
#creating a bar plot to group each company's average laptop price
avg_price = cleaned_laptop.groupby('company')['price'].mean().sort_values()

plt.figure(figsize=(10, 6))
avg_price.plot(kind='bar', color='green', edgecolor='black')
plt.title('Average Laptop Price by Company')
plt.xlabel('Company')
plt.ylabel('Average Price')
plt.show()
```

```python
#a bar plot showing the count of each laptoptype in the company
type_counts = cleaned_laptop['typename'].value_counts()
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='orange', edgecolor='black')
plt.title('Number of Laptops by Type')
plt.xlabel('Laptop Type')
plt.ylabel('Count')
plt.show()
```

```python
#a bar plot showing the count the cpu brand name
type_counts = cleaned_laptop['cpu_brand'].value_counts()
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='red', edgecolor='black')
plt.title('Number of Cpu brands')
plt.xlabel('Cpu brands')
plt.ylabel('Count')
plt.show()
```

```python
#a bar plot showing the count the gpu brand name
type_counts = cleaned_laptop['gpu_brand'].value_counts()
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='violet', edgecolor='black')
plt.title('Number of gpu brands')
plt.xlabel('Gpu brands')
plt.ylabel('Count')
plt.show()
```

```python
# a pie chart showing the count of the number of operating systems
os_counts = cleaned_laptop['operating_system'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(os_counts, labels=os_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Operating Systems')
plt.show()
```
