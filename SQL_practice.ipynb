```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Book_Title': ['Ruthless People', 'ACOTAR', 'Fourth Wing', 'Beautiful Venom', 'The wrong quaterback'],
    'Book_Author': ['J.J McAvoy', 'Sarah J. Maas', 'Rebecca Yarros', 'Rina kent', 'Alex Michaelides'],
    'Book_Genre': ['Mafia', 'Fantasy', 'Fantasy', 'Dark Romace', 'Billionaire Romance'],
    'Price': [10.99, 12.99, 14.99, 16.99, 18.99],
    'Units_Sold': [100, 200, 300, 400, None],
    'Book_ID': [1, 2, 3, 4, 5],
    'Quantity': [10, 20, 30, 40, 50],
    'Author_id': [1, 2, 3, 4, 5],
    'Book_Rating': [4.2, 4.7, 4.5, 4.9, 4.6]
}

books = pd.DataFrame(data)
```

```python
import sqlite3

conn = sqlite3.connect("data.db")

books.to_sql("data", conn, if_exists="replace", index=False)
```
