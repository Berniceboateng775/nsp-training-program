```python
import pandas as pd
import numpy as np
import sqlite3
```

```python
conn = sqlite3.connect("bookstore.db")
#books data
books_data = {
    'book_id': [1, 2, 3, 4, 5 ],
    'book_title': ['Ruthless People', 'Beautiful Venom', 'ACOTAR', 'Fourth Wing', 'The Wrong Quaterback'],
    'book_genre': ['Mafia', 'Dark Romance', 'Fantasy', 'Fantasy', 'Billionaire Romance'],
    'price': [19.99, 14.99, 29.99, 19.99, 14.99],
    'units_sold': [120, 200, 150, 300, None],
    'quantity': [10, 20, 15, 25, 5],
    'author_id': [1, 2, 3, 4, 5],
    'book_rating': [4.2, 4.5, 4.8, 4.1, 4.3]
}
books = pd.DataFrame(books_data)
books.head()
```

```python
#Authors Data
authors_data = {
    'author_id': [1, 2, 3, 4, 5, 6],
    'author_name': ['J.J McAvoy', 'Rina Kent', 'Sarah J. Maas', 'Rebecca Yarros',  'C.R Jane', 'Penelope Ward'],
    'country': ['Canada', 'USA', 'USA', 'uk', 'Cyprus', 'USA' ],
    'birth_year': [1986, 1980, 1976, 1972, 1970, 1968]
}
authors = pd.DataFrame(authors_data)
authors.head()
```

```python
#orders data
orders_data = {
    'order_id': [1, 2, 3, 4, 5, 6, 7],
    'book_id': [2, 3, 1, 3, 4, 2, 5],
    'customer_name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'order_date': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05', '2023-01-06', '2023-01-07'],
    'quantity_ordered': [2, 1, 3, 2, 1, 4, 2]
}
orders = pd.DataFrame(orders_data)
orders.head()
```

```python
#saving all to SQLite database
books.to_sql("books", conn, if_exists="replace", index=False)
authors.to_sql("authors", conn, if_exists="replace", index=False)
orders.to_sql("orders", conn, if_exists="replace", index=False)

print("All tables created successfully!")
```

**Table Relationship**  
*1. books.author_id  →  authors.author_id  (Foreign Key)*             
*2. orders.book_id   →  books.book_id      (Foreign Key)*

**Basic SELECT - Retrieving Data**

```python
#selecting all data from the books table
pd.read_sql("SELECT * FROM books;", conn)
```

```python
#selecting specific columns only
pd.read_sql("SELECT book_title, price, book_rating FROM books;", conn)
```

```python
#using aliases to rename columns in output
pd.read_sql("""
SELECT 
book_title AS 'Title', 
price AS 'Price ($)', 
book_rating AS 'Rating' 
FROM books;
""", conn)
```

**WHERE - Filtering Data**

```python
#filtering by a single condition
pd.read_sql("SELECT * FROM books WHERE price > 15;", conn)
```

```python
#multiple conditions with AND
pd.read_sql("""
SELECT * FROM books
WHERE price > 15 
AND book_rating > 4.2;
""", conn)
```

```python
#using OR for alternative conditions
pd.read_sql("""
SELECT * FROM books
WHERE book_genre = 'Fantasy'
OR book_genre = 'Mafia';
""", conn
)
```

```python
#using IN for multiple values 
pd.read_sql("""
SELECT * FROM books 
WHERE book_genre IN ('Fantasy', 'Mafia');
""", conn)
```

```python
#using BETWEEN for ranges
pd.read_sql("""
SELECT * FROM books 
WHERE price BETWEEN 14 AND 20;
""", conn)
```

```python
#pattern matching with LIKE
# % = any characters, _ = single character
pd.read_sql("""
SELECT * FROM books 
WHERE book_title LIKE '%Wing%';
""", conn)
```

```python
#pattern matching with LIKE
# % = any characters, _ = single character
pd.read_sql("""
SELECT * FROM books 
WHERE book_title LIKE '%K%';
""", conn)
```

**ORDER BY - Sorting**

```python
#sorting by price ascending 
pd.read_sql("SELECT * FROM books ORDER BY price ASC;", conn)
```

```python
#sorting by price descending 
pd.read_sql("SELECT * FROM books ORDER BY price DESC;", conn)
```

```python
#sorting by  multiple columns
pd.read_sql("""
SELECT * FROM books 
ORDER BY book_genre, book_rating DESC;
""", conn)
```

**LIMIT - Restricting Results**

```python
#getting only top 3 results
pd.read_sql("SELECT * FROM books ORDER BY price DESC LIMIT 3;", conn)
```

**Working with NULL values**

```python
#finding rows WHERE a column IS NULL
pd.read_sql("""
SELECT * FROM books 
WHERE units_sold IS NULL;
""", conn)
```

```python
#finding rows WHERE a column IS NOT NULL
pd.read_sql("""
SELECT * FROM books 
WHERE units_sold IS NOT NULL;
""", conn)
```

```python

# fyi - this will NOT work :
pd.read_sql("SELECT * FROM books WHERE units_sold = NULL;", conn)
# Always use IS NULL or IS NOT NULL!
```

```python
#Using COALESCE to replace NULL with a default value
pd.read_sql("""
SELECT 
book_title,
units_sold,
COALESCE(units_sold, 0) AS units_sold_cleaned
FROM books;
""", conn)
```

**Aggregate Functions & GROUP BY**

```python
#count all rows
pd.read_sql("SELECT COUNT(*) AS total_books FROM books;", conn)
```

```python
#SUM - Total of a column
pd.read_sql("SELECT SUM(units_sold) AS total_units FROM books;", conn)
```

```python
#Average value
pd.read_sql("SELECT AVG(price) AS avg_price FROM books;", conn)
```

```python
#min  and max
pd.read_sql("""
SELECT 
MIN(price) AS cheapest,
MAX(price) AS most_expensive
FROM books;
""", conn)
```

```python
#round decimal values
pd.read_sql("SELECT ROUND(AVG(price), 1) AS avg_price FROM books;", conn)
```

**GROUP BY - Grouping Data**

```python
#counting books per genre
pd.read_sql("""
SELECT 
book_genre,
COUNT(*) AS book_count
FROM books
GROUP BY book_genre;
""", conn)
```

```python
#Average price per genre
pd.read_sql("""
SELECT 
book_genre,
ROUND(AVG(price), 2) AS avg_price
FROM books
GROUP BY book_genre;
""", conn)
```

```python
#Total revenue per genre (price * units_sold)
pd.read_sql("""
SELECT 
book_genre,
SUM(price * units_sold) AS revenue
FROM books
GROUP BY book_genre
ORDER BY revenue DESC;
""", conn)
```

```python
#using HAVING to filter groups (like WHERE but for groups)
pd.read_sql("""
SELECT 
book_genre,
COUNT(*) AS book_count,
ROUND(AVG(book_rating), 2) AS avg_rating
FROM books
GROUP BY book_genre
HAVING COUNT(*) > 1;
""", conn)
```

**INNER JOIN - Only matching rows from BOTH tables**

```python
#getting books with their author names
#only returns books that HAVE a matching author
pd.read_sql("""
SELECT 
b.book_title,
b.price,
a.author_name,
a.country
FROM books b
INNER JOIN authors a ON b.author_id = a.author_id;
""", conn)
```

```python
#getting orders with book details
pd.read_sql("""
SELECT 
o.order_id,
o.customer_name,
b.book_title,
b.price,
o.quantity_ordered,
(b.price * o.quantity_ordered) AS total_cost
FROM orders o
INNER JOIN books b ON o.book_id = b.book_id;
""", conn)
```

**LEFT JOIN - All rows from LEFT table + matching from right**

```python
#getting ALL authors, even those without books
#Penelope Ward (author_id=6) has no books - she will still show!
pd.read_sql("""
SELECT 
a.author_name,
a.country,
b.book_title,
b.price
FROM authors a
LEFT JOIN books b ON a.author_id = b.author_id;
""", conn)
```

```python
#finding authors who have NO books (NULL book_title)
pd.read_sql("""
SELECT 
a.author_name,
a.country
FROM authors a
LEFT JOIN books b ON a.author_id = b.author_id
WHERE b.book_id IS NULL;
""", conn)
```

**Multiple JOINs - Connecting 3+ tables**

```python
#getting orders with book AND author info
pd.read_sql("""
SELECT 
o.order_id,
o.customer_name,
o.order_date,
b.book_title,
a.author_name,
b.price,
o.quantity_ordered
FROM orders o
INNER JOIN books b ON o.book_id = b.book_id
INNER JOIN authors a ON b.author_id = a.author_id
ORDER BY o.order_date;
""", conn)
```

**JOINs with GROUP BY**

```python
#total quantity ordered per book
pd.read_sql("""
SELECT 
b.book_title,
SUM(o.quantity_ordered) AS total_ordered
FROM books b
LEFT JOIN orders o ON b.book_id = o.book_id
GROUP BY b.book_title
ORDER BY total_ordered DESC;
""", conn)
```

```python
#revenue per author
pd.read_sql("""
SELECT 
a.author_name,
a.country,
SUM(b.price * o.quantity_ordered) AS total_revenue
FROM authors a
LEFT JOIN books b ON a.author_id = b.author_id
LEFT JOIN orders o ON b.book_id = o.book_id
GROUP BY a.author_name, a.country
ORDER BY total_revenue DESC;
""", conn)
```

```python
#books which has never been ordered
pd.read_sql("""
SELECT 
b.book_title,
b.price
FROM books b
LEFT JOIN orders o ON b.book_id = o.book_id
WHERE o.order_id IS NULL;
""", conn)
```

**FULL OUTER JOIN**

```python
#getting ALL books and ALL authors (using UNION)
pd.read_sql("""
SELECT a.author_name, b.book_title
FROM authors a
LEFT JOIN books b ON a.author_id = b.author_id

UNION

SELECT a.author_name, b.book_title
FROM books b
LEFT JOIN authors a ON a.author_id = b.author_id;
""", conn)
```
